```python
# Imports
import pandas as pd
import matplotlib.pyplot as plt

# Import the data into Pandas dataframe
data = pd.read_csv("API_testing_project.csv")
```

```python
data.head()
```

```python
data["result"].value_counts(normalize=True)
```

```python
data[data["result"] == "FAIL"]
```

```python
avg_response_time = round(data["response_time_ms"].mean(), 2)

print(f"Average API response time: {avg_response_time} milliseconds")
```

```python
tests_valid = data[data["result"] == "PASS"]

avg_time_valid = round(tests_valid["response_time_ms"].mean(), 2)

print(f"Average API reponse time among valid test cases: {avg_time_valid}")
```

```python
# Fastest and slowest response times
fastest = min(data["response_time_ms"])
slowest = max(data["response_time_ms"])

print(f"Fastest time: {fastest}")
print(f"Slowest time: {slowest}")
```

```python
# Fastest response
fastest_test = data.loc[data["response_time_ms"] == fastest, "category"].iloc[0]
slowest_test = data.loc[data["response_time_ms"] == slowest, "category"].iloc[0]

print(f"Fastest test: {fastest_test}")
print(f"Slowest test: {slowest_test}")
```

```python
# Average response time per category?

avg_res_time = data.groupby(["category"])["response_time_ms"].mean()
avg_res_time
```
