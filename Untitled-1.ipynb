```python
test_runs = [
    {"tester": "Cameron", "status": "passed"},
    {"tester": "Travis", "status": "failed"},
    {"tester": "Cameron", "status": "failed"},
    {"tester": "Erik", "status": "passed"},
    {"tester": "Travis", "status": "failed"},
    {"tester": "Cameron", "status": "failed"},
    {"tester": "Erik", "status": "failed"}
]
```

```python
def most_failures(test_runs):
    failed_dict = {}
    for test in test_runs:
        if test["status"] == "failed" and test["tester"] in failed_dict:
            failed_dict[test["tester"]] += 1
        elif test["status"] == "failed":
            failed_dict[test["tester"]] = 1
    return max(failed_dict, key=failed_dict.get)
```

```python
most_failures(test_runs)
```
